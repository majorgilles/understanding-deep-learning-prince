# Understanding isocontours in the loss plot

![Figure 2.3: three-dimensional loss surface and two-dimensional isocontours](assets/figure-2-3-loss-isocontours.png)

> **Figure source:** Figure 2.3 from [*Understanding Deep Learning*](../chapter-02-supervised-learning.pdf), book page 21, by Simon J. D. Prince (MIT Press). The supplied screenshot is included unchanged for noncommercial study under [CC BY-NC-ND 4.0](../../../LICENSES/CC-BY-NC-ND-4.0.txt).

The correct term is **isocontours** (also called **contour lines** or **level curves**):

- **iso** means ‘equal’ or ‘the same’;
- a **contour** is a line joining points with the same value.

Therefore, an isocontour of the loss function connects every pair of model parameters that produces the **same loss**.

## What does one point in the plot represent?

The linear regression model is

$$y = \phi_0 + \phi_1 x,$$

where $\phi_0$ is the intercept and $\phi_1$ is the slope. A single point $(\phi_0, \phi_1)$ in figure 2.3b therefore represents **one complete regression line**, not one training example. Changing position in this plot means choosing a different intercept and slope.

For every possible parameter pair, the book computes a loss such as the sum of squared errors:

$$L(\phi_0, \phi_1) = \sum_i \left(\phi_0 + \phi_1x_i-y_i\right)^2.$$

This gives three quantities: intercept, slope, and loss.

## Connecting panels (a) and (b)

Panel **(a)** shows the loss as the height of a three-dimensional bowl:

- moving along one ground axis changes the intercept $\phi_0$;
- moving along the other ground axis changes the slope $\phi_1$;
- the height of the bowl is the loss $L(\phi_0,\phi_1)$.

Panel **(b)** looks straight down at that same bowl. Imagine cutting the bowl with several horizontal planes. Each plane intersects the bowl in a curve of constant height. Viewed from above, those curves are the gray elliptical isocontours. In other words, each ellipse joins slope/intercept combinations with equal loss.

This is exactly like a topographic map:

- the mountain's height corresponds to loss;
- a contour line connects locations at the same elevation;
- here we want to reach the **lowest valley**, because lower loss means a better-fitting model.

## How to read the ellipses

Suppose one gray ellipse represents $L=10$. Every parameter pair on that ellipse has loss 10, even though the individual slopes and intercepts differ.

- Moving **along** an isocontour keeps the loss constant.
- Crossing toward the center moves to contours with smaller loss.
- Crossing outward moves to contours with larger loss.
- Closely spaced contours mean the loss changes quickly in that direction—a steep part of the bowl.
- Widely spaced contours mean the loss changes more slowly.

The background brightness encodes the same information: brighter regions have larger loss and darker regions have smaller loss. The green circle marks the parameters of the best of the three illustrated regression lines, so it lies in a lower-loss region than the cyan and white circles.

## Why are the contours tilted ellipses?

For linear regression with squared-error loss, the loss is a quadratic function of the parameters, so horizontal slices through its bowl are ellipses. Their tilt and elongation show that slope and intercept interact: increasing the intercept can sometimes be partly compensated for by changing the slope.

The long direction of an ellipse is a direction in which several parameter combinations have similar loss. The narrow direction is one in which a small parameter change affects the loss more strongly.

## Connection to gradient descent

The gradient points perpendicular to an isocontour toward the fastest increase in loss. Gradient descent moves in the opposite direction, crossing contours from higher loss to lower loss until it approaches the minimum at the bottom of the bowl.

## One-sentence summary

An isocontour is a line on the slope/intercept map connecting all regression models that make different parameter choices but have the same loss.
