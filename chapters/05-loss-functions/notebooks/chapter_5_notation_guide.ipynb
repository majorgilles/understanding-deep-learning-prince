# Chapter 5 notation companion - Loss functions

This Markdown-only companion follows [Chapter 5: Loss functions](../chapter-05-loss-functions.pdf), book pages 56-76. Its purpose is to explain the formulas, mathematical symbols, and probability notation used in the chapter.

The chapter's central pipeline is:

$$
x_i \longrightarrow f[x_i,\phi]=\theta_i \longrightarrow \Pr(y_i\mid\theta_i) \longrightarrow -\log\Pr(y_i\mid\theta_i).
$$

Read it from left to right: training input $x_i$ enters a model with parameters $\phi$; the model produces parameters $\theta_i$ for an output distribution; that distribution assigns probability or probability density to the observed target $y_i$; the negative log of this value becomes the example's loss.

> **Book convention:** square brackets denote function arguments. For example, $f[x,\phi]$ has the same meaning as the more common $f(x,\phi)$.

## Domains used in Chapter 5

- $y\in\mathbb{R}$: $y$ is any real number.
- $y\in\mathbb{R}^{K}$: $y$ is a vector of $K$ real numbers.
- $y\in\mathbb{R}^{+}$: $y$ is positive.
- $y\in\{0,1\}$: $y$ is one of two labels.
- $y\in\{1,2,\ldots,K\}$: $y$ is one of $K$ class labels.
- $y\in\{0,1,2,\ldots\}$: $y$ is a nonnegative integer count.
- $y\in(-\pi,\pi]$: $y$ is an angle on a circular domain.

# A beginner introduction to integrals

An integral is a **continuous sum**. A normal sum adds separate values:

$$
\sum_{i=1}^{I} a_i=a_1+a_2+\cdots+a_I.
$$

An integral adds infinitely many tiny contributions across a continuous interval:

$$
\int_a^b g(y)\,dy.
$$

Read this as: 'integrate $g(y)$ with respect to $y$, from $a$ to $b$.'

| Part | Meaning |
|---|---|
| $\int$ | Continuous summation symbol |
| $a,b$ | Lower and upper bounds |
| $g(y)$ | Integrand: the quantity being accumulated |
| $dy$ | The accumulation variable is $y$; imagine an extremely thin width in the $y$ direction |

Geometrically, if $g(y)\geq0$, the integral is the area under the curve. For a probability density $p(y)$, total area must equal one:

$$
\int_{-\infty}^{\infty}p(y)\,dy=1.
$$

Probability over an interval is area over that interval:

$$
\Pr(a\leq y\leq b)=\int_a^b p(y)\,dy.
$$

For discrete outcomes, use a sum; for continuous outcomes, use an integral. Chapter 5 later integrates probability densities when defining KL divergence and cross-entropy.

> **Dummy variable:** in $\int p(z)\,dz$, the letter $z$ only names the axis being accumulated. Replacing every $z$ with $y$ leaves the value unchanged.

## The critical distinction: phi versus theta

- $\phi$ belongs to the neural network and is shared across all examples. Training changes $\phi$.
- $\theta_i$ belongs to the predicted distribution for example $i$. It changes with the input because $\theta_i=f[x_i,\phi]$.

For regression, $\theta_i$ may contain a mean and variance. For binary classification, it contains a Bernoulli probability. For multiclass classification, it contains one probability per class.

# Probability, conditional probability, density, and likelihood

$$
\Pr(y\mid x)
$$

is read 'the probability of $y$ given $x$.' After observing $x$, it describes how plausible each possible output $y$ is.

The same expression has two viewpoints:

1. Hold parameters fixed and vary $y$: it is a **probability distribution** over outputs.
2. Hold observed $y$ fixed and vary parameters: it is a **likelihood** measuring which parameters explain the observation well.

Likelihood is not a different equation. It is the same equation treated as a function of different variables. A probability distribution normalizes over possible outcomes; a likelihood does not generally normalize over parameter values.

> **Continuous-output warning:** when $y$ is real-valued, $\Pr(y\mid\theta)$ in this chapter usually denotes a probability **density**. A density can exceed one at a point; its integral over the full domain must equal one.

# Summation, products, and independence

Summation adds one term for every index value:

$$
\sum_{i=1}^{I}a_i=a_1+a_2+\cdots+a_I.
$$

Product notation multiplies one term for every index value:

$$
\prod_{i=1}^{I}p_i=p_1p_2\cdots p_I.
$$

The chapter assumes training outputs are conditionally independent given their inputs. This lets the joint likelihood factor into per-example terms:

$$
\Pr(y_1,\ldots,y_I\mid x_1,\ldots,x_I)=\prod_{i=1}^{I}\Pr(y_i\mid x_i).
$$

Independence is a modeling assumption. It is what allows the dataset score to be assembled example by example.

# argmax, argmin, hats, and maximum likelihood

- $\max_{\phi}g(\phi)$ returns the largest **value** of $g$.
- $\arg\max_{\phi}g(\phi)$ returns the value of $\phi$ **where** that maximum occurs.
- $\arg\min$ similarly returns where a function is smallest.

Maximum likelihood is written

$$
\hat{\phi}=\arg\max_{\phi}\prod_{i=1}^{I}\Pr(y_i\mid f[x_i,\phi]).
$$

Read it as: 'phi hat is the choice of network parameters phi that makes the observed targets collectively most likely.' The hat means estimated, fitted, or selected value.

# Log-likelihood and negative log-likelihood

The logarithm is monotonically increasing, so it does not move the location of a maximum. It also changes products into sums:

$$
\log\left(\prod_i p_i\right)=\sum_i\log p_i.
$$

This avoids numerical underflow from multiplying many small probabilities. Multiplying by minus one converts maximization into the conventional minimization form:

$$
\mathcal{L}[\phi]=-\sum_{i=1}^{I}\log\Pr(y_i\mid f[x_i,\phi]).
$$

This is the **negative log-likelihood** (NLL). A likely observation gets a small penalty: $-\log(0.9)\approx0.105$. An unlikely observation gets a large penalty: $-\log(0.1)\approx2.303$.

# The Chapter 5 loss-construction recipe

1. Identify the domain of $y$: real, binary, categorical, circular, count, and so on.
2. Choose a probability distribution defined on that domain.
3. Let the network predict one or more distribution parameters: $\theta_i=f[x_i,\phi]$.
4. If needed, transform raw network outputs so the parameters satisfy constraints.
5. Score each observed target using $\Pr(y_i\mid\theta_i)$.
6. Train by minimizing $-\sum_i\log\Pr(y_i\mid\theta_i)$.
7. At inference, return the full distribution or a point estimate such as its mode.

Least squares, binary cross-entropy, and multiclass cross-entropy are all instances of this recipe with different output distributions.

# Normal distribution notation for regression

For $y\in\mathbb{R}$, Chapter 5 chooses a univariate normal distribution:

$$
\Pr(y\mid\mu,\sigma^2)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left[-\frac{(y-\mu)^2}{2\sigma^2}\right].
$$

| Symbol | Meaning |
|---|---|
| $\mu$ (mu) | Mean; center and peak location |
| $\sigma^2$ (sigma squared) | Variance; squared spread or uncertainty |
| $\sigma=\sqrt{\sigma^2}$ | Standard deviation; spread in the units of $y$ |
| $y-\mu$ | Residual; observation minus predicted mean |
| $\exp[z]=e^z$ | Exponential function |
| $\pi$ | Circle constant, approximately $3.14159$ |

The network predicts the mean $\mu_i=f[x_i,\phi]$. The variance may be fixed, learned as one shared value, or predicted separately for each input.

# Why normal NLL becomes least squares

After substituting $\mu_i=f[x_i,\phi]$ and taking the negative logarithm, normal NLL has the structure

$$
\mathcal{L}[\phi]=C+\frac{1}{2\sigma^2}\sum_{i=1}^{I}(y_i-f[x_i,\phi])^2,
$$

where $C$ is independent of $\phi$. If $\sigma^2$ is fixed and positive, neither $C$ nor the positive factor $1/(2\sigma^2)$ changes the location of the minimum. Hence

$$
\hat{\phi}=\arg\min_{\phi}\sum_{i=1}^{I}(y_i-f[x_i,\phi])^2.
$$

This is least-squares loss. The simplified expression does not have the same numerical value as full NLL; it has the same optimizer under the fixed-variance assumptions.

For inference, the normal distribution's mode is its mean, so $\hat y=f[x,\hat{\phi}]$.

# Homoscedastic and heteroscedastic uncertainty

**Homoscedastic** means the conditional variance is constant:

$$
\mu_i=f[x_i,\phi],\qquad \sigma_i^2=\sigma^2.
$$

**Heteroscedastic** means uncertainty changes with the input. The network produces two outputs:

$$
\mu_i=f_1[x_i,\phi],\qquad \sigma_i^2=f_2[x_i,\phi]^2.
$$

$f_1$ and $f_2$ denote two output components of one model. Squaring enforces $\sigma_i^2\geq0$. When variance is learned, its logarithmic term cannot be discarded because it depends on the model parameters.

# Binary classification: Bernoulli, sigmoid, and BCE

For $y\in\{0,1\}$, the Bernoulli parameter is

$$
\lambda=\Pr(y=1),\qquad 1-\lambda=\Pr(y=0).
$$

Both labels fit into one formula:

$$
\Pr(y\mid\lambda)=(1-\lambda)^{1-y}\lambda^y.
$$

The exponents act as switches. For $y=1$, the formula reduces to $\lambda$; for $y=0$, it reduces to $1-\lambda$.

A raw network score $z=f[x,\phi]$ can be any real number. The sigmoid maps it into $[0,1]$:

$$
\lambda=\operatorname{sig}(z)=\frac{1}{1+\exp(-z)}.
$$

The Bernoulli NLL is binary cross-entropy:

$$
\mathcal{L}[\phi]=-\sum_{i=1}^{I}\left[(1-y_i)\log(1-\lambda_i)+y_i\log\lambda_i\right].
$$

For $y_i=1$, the example loss is $-\log\lambda_i$. For $y_i=0$, it is $-\log(1-\lambda_i)$. The unrestricted score $z$ is commonly called a **logit**.

# Multiclass classification: categorical and softmax

For $K$ classes, $y\in\{1,2,\ldots,K\}$. A categorical distribution has probabilities

$$
\lambda_1,\ldots,\lambda_K,\qquad 0\leq\lambda_k\leq1,\qquad \sum_{k=1}^{K}\lambda_k=1.
$$

The network produces $K$ unrestricted logits $z_k=f_k[x,\phi]$. Softmax converts them to probabilities:

$$
\operatorname{softmax}_k(z)=\frac{\exp(z_k)}{\sum_{k'=1}^{K}\exp(z_{k'})}.
$$

$k$ identifies the class being computed. $k'$ is a dummy index visiting every class in the denominator.

For example $i$, softmax component $y_i$ is the probability assigned to the correct class. Multiclass cross-entropy is

$$
\mathcal{L}[\phi]=-\sum_{i=1}^{I}\log\left(\operatorname{softmax}_{y_i}(f[x_i,\phi])\right).
$$

For a point prediction, choose $\hat y=\arg\max_k\lambda_k$.

# Multiple-output notation

For vector targets, $y_{id}$ means component $d$ of training example $i$. Under conditional independence across components:

$$
\Pr(\mathbf y\mid f[x,\phi])=\prod_d\Pr(y_d\mid f_d[x,\phi]).
$$

Negative logs turn the product into an additional sum:

$$
\mathcal{L}[\phi]=-\sum_{i=1}^{I}\sum_d\log\Pr(y_{id}\mid f_d[x_i,\phi]).
$$

$d$ indexes output components. $f_d$ means the set of model outputs needed to parameterize the distribution for component $d$; a categorical component requires $K$ logits rather than one number.

# Cross-entropy, KL divergence, and Dirac deltas

> **Beginner path:** If the integral notation is unfamiliar, first work through [`chapter_05_cross_entropy_and_integrals_from_scratch.ipynb`](chapter_05_cross_entropy_and_integrals_from_scratch.ipynb), which builds definite integrals from rectangle sums before deriving cross-entropy, KL divergence, Dirac selector spikes, and Equations 5.27–5.31.

Section 5.7 uses $q(y)$ for the empirical data distribution and $p(y)=\Pr(y\mid\theta)$ for the model distribution. KL divergence is

$$
D_{KL}(q\parallel p)=\int q(y)\log\frac{q(y)}{p(y)}\,dy
=\int q(y)\log q(y)\,dy-\int q(y)\log p(y)\,dy.
$$

$q\parallel p$ is read 'q relative to p.' The double bar is not the conditional-probability bar. KL divergence is ordered, so generally $D_{KL}(q\parallel p)\neq D_{KL}(p\parallel q)$.

Cross-entropy is

$$
H(q,p)=-\int q(y)\log p(y)\,dy.
$$

The empirical continuous distribution is represented as point masses:

$$
q(y)=\frac{1}{I}\sum_{i=1}^{I}\delta(y-y_i).
$$

$\delta(y-y_i)$ is a Dirac delta centered at $y_i$. It is not an ordinary finite-height function; it is defined by the selection property

$$
\int\delta(y-y_i)g(y)\,dy=g(y_i).
$$

Substituting $q$ into cross-entropy selects the model log-density at every observed point, producing $-(1/I)\sum_i\log p(y_i)$. The fixed factor $1/I$ can be removed without moving the optimum. This is why empirical cross-entropy and NLL are equivalent in the chapter.

# From prediction task to loss

| Target domain | Distribution | Constraint transform | Common loss | Point estimate |
|---|---|---|---|---|
| $y\in\mathbb{R}$ | Normal | Identity for mean; positive transform for variance | Least squares when variance is fixed | Mean |
| $y\in\{0,1\}$ | Bernoulli | Sigmoid | Binary cross-entropy | Threshold or mode |
| $y\in\{1,\ldots,K\}$ | Categorical | Softmax | Multiclass cross-entropy | Highest-probability class |
| $y\in\{0,1,2,\ldots\}$ | Poisson | Positive transform for rate | Poisson NLL | Distribution-dependent count estimate |
| $y\in(-\pi,\pi]$ | von Mises | Circular mean and positive concentration | von Mises NLL | Mean direction/mode |

The transformation is essential: it converts unrestricted network outputs into valid distribution parameters.

# Common notation traps and reading checklist

1. The network often predicts distribution parameters, not the final point prediction directly.
2. $\phi$ is shared model state; $\theta_i$ is the distribution description for one example.
3. $\arg\max$ returns a location, not the maximum height.
4. A hat marks an estimate.
5. For continuous outputs, $\Pr$ usually denotes density.
6. A term can be dropped from an optimization only when it is independent of the optimized parameter, or is a fixed positive scale.
7. Sum and mean losses have the same optimizer but different scales.
8. Independence is an assumption that enables products to factor.
9. Sigmoid and softmax consume unrestricted logits and produce probabilities.
10. Cross-entropy and NLL begin from different viewpoints but reduce to the same empirical criterion here.

## Five questions for reading any Chapter 5 formula

1. What is the domain of $y$?
2. Which distribution is being used?
3. Which symbols are model parameters, and which are distribution parameters?
4. What constraints must the distribution parameters satisfy?
5. Is the expression a probability, likelihood, log-likelihood, loss, or fitted estimate?
